```python
# Code to calculate the thickness of a paper
def folding_paper(thickness: float, folding_times: int) -> float :
  ''' Function that calculates the thickness of a sheet of paper in meters.


  Detailed description of the function


  :param thichness: thickness of the paper to be foldes
  :param folding_times: number of times that the paper will be folded
  :raise: TypeError, if the argument thickness is not a type float
  :raise: TyperError, if the argument folding_times is not a type int
  :raise: ValueError, if the argument thickness is not positive
  :raise: ValueError, if the argument folding_times is not positive
  :print: The thickness of the folded paper in meters

  '''
  if not isinstance(thickness, float):
    raise TypeError(f'thickness variable should be type float, not {type(thickness)}')
  if not isinstance(folding_times, int):
    raise TypeError(f'folding_times should be type int, not {type(folding_times)}')
  if not thickness > 0:
    raise ValueError("The number should be positive")
  if not folding_times > 0:
    raise ValueError("The number should be positive")
  folded_thickness = thickness*2**folding_times
  return folded_thickness
print(f"Thickness: {folding_paper(0.00008, 43)} meters")
```
