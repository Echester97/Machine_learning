**PROBLEM 1** Creating random numbers

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)
mean = [-3, 0]
cov = [[1.0, 0.8],[0.8, 1.0]]
mult = np.random.multivariate_normal(mean, cov, size=500)
```

**PROBLEM 2** Visualization by scatter plot

```python
x1 = mult[:, 0]
x2 = mult[:, 1]
plt.scatter(x1, x2, alpha=0.5, cmap='coolwarm', edgecolors='k')
plt.xlabel('x1')
plt.ylabel('x2')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

**PROBLEM 3** Visualization by histogram

```python
fig, ax = plt.subplots(2)
ax[0].set_xlabel('x1')
ax[0].set_xlim(-5.6, 2.5)
ax[0].set_ylabel('Frequency')
ax[0].hist(x1, bins=30, density=True, alpha=0.5, color='blue', label='x1', edgecolor='black')
ax[1].set_xlabel('x2')
ax[1].set_xlim(-5.6, 2.5)
ax[1].set_ylabel('Frequency')
ax[1].hist(x2, bins=30, density=True, alpha=0.5, color='red', label='x2', edgecolor='black')
plt.show()
```

**PROBLEM 4** Addition of data

```python
np.random.seed(0)
mean = [-3, 0]
cov = [[1.0, 0.8],[0.8, 1.0]]
mult = np.random.multivariate_normal(mean, cov, size=500)

mean = [0, -3]
cov = [[1.0, 0.8],[0.8, 1.0]]
mult_2 = np.random.multivariate_normal(mean, cov, size=500)

x1 = mult[:, 0]
x2 = mult[:, 1]
x3 = mult_2[:, 0]
x4 = mult_2[:, 1]

fig,ax = plt.subplots()
ax.set_xlabel('x1')
ax.set_ylabel('x2')
ax.scatter(x1, x2, alpha=0.5, cmap='coolwarm', edgecolors='k', label='0')
ax.scatter(x3, x4, alpha=0.5, cmap='inferno', edgecolors='k', label='1')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()




```

**PROBLEM 5** Data combination

```python
combined_mult = np.concatenate((mult, mult_2), axis=0)
combined_mult.shape
```

**PROBLEM 6** Labeling

```python
label_1 = np.zeros((500,1))
label_4 = np.ones((500,1))
label_4.shape
mult_labeled = np.concatenate((mult, label_1), axis=1)
mult_2_labeled = np.concatenate((mult_2, label_4), axis=1)
combined_mult_labeled = np.concatenate((mult_labeled, mult_2_labeled), axis=0)
print(combined_mult_labeled.shape)
print('\n')
print(combined_mult_labeled)
```
