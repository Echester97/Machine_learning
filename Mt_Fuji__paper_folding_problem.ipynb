**PROBLEM 1 How many times to fold paper to exceed the height of Mt.Fuji**

```python
thickness = 0.00008 # in meter
mt_fuji = 3776 # in meter
fold_times = 0
folded_thickness = thickness
while folded_thickness <= mt_fuji:
  folded_thickness *= 2
  fold_times += 1

print(f'To exceed the Mt.Fuji you need to fold a paper {fold_times} times')
```

**PROBLEM 2**

```python
def times_to_fold(distance:float, thickness:float =0.00008)-> int:
  '''
  A function that calculates the times that it is needed to fold a paper to reach a certain distance.

  Parameters

  :distance(float): Distance to reach
  :thickness(float): Thickness of the paper (default = 0.00008)

  Returns

  - folding_times(float): Number of times that the paper is folded

  '''
  folding_times = 0
  current_thickness = thickness
  while current_thickness <= distance:
    current_thickness *= 2
    folding_times += 1
  return folding_times

distance_to_proxima_centauri = 3.991e16 # in meters

# Calculate the folding times

number_of_times = times_to_fold(distance_to_proxima_centauri)

print(f'To reach Proxima Centauri you need to fold a paper {number_of_times} times')


```
