**PROBLEM 1 Number of grains of rice on the 100th day**

```python
import numpy as np
import matplotlib.pyplot as plt
# Calculate how many grains you receive in total on the 100th day
days = 100
total_rice_grains = 1*2**days
print(f'Number of grains received: {total_rice_grains}')

##
# Define the range of days
number_of_days = np.arange(0, 101, 1)

#Calculate the number of rice grains for each day
number_of_rice = 2.0**(number_of_days -1)

# Calculate the total number of rice grains from day 1 to day 100
rice_grain = 1
number_of_rices_values = []
total_number_of_rice = rice_grain
number_of_rices_values.append(total_number_of_rice)

for i in range(100):
    total_number_of_rice *=2
    number_of_rices_values.append(total_number_of_rice)
print(f"Number of rice on the 100th day:{number_of_rices_values[-2]}")

#Plot the results
fig, ax = plt.subplots()
ax.plot(number_of_days, number_of_rice, color='blue', lw=2, label="Rice grains for each day")
ax.plot(number_of_days, number_of_rices_values, linestyle='--', color='red', lw=2, label="Rice growth")
ax.set_xlabel("Number of Days")
ax.set_ylabel("Number of grains")
ax.set_yscale('linear')
ax.set_title("Number of grains of rice on the 100th day")
ax.grid(True)
ax.legend()
plt.show()
```

**PROBLEM 2 Number of rice grains outside of the 100th day**

```python
def compute_sorori_shinzaemon(n_days:int):
    """A function that calculates the number of rice grains per day in the rice anecdote of Shinzaemon Sorori.

    Parameteres
    -----------
    n_days: int
        Number of days to get rice (default : 100)

    Returns
    -------
    list_n_grains : list
        List of the number of rice you get one day (list length equals n_days)
    list_total_grains : list
        List of total number of rice received by a certain day (list length is equal to n_days)
    """
    list_n_grains = []
    list_total_grains = []
    grain = 1

    for day in range(n_days):
        if day == 0:
            list_n_grains.append(grain)
            list_total_grains.append(grain)
        else:
            grain = grain*2
            list_n_grains.append(grain)
            list_total_grains.append(list_total_grains[-1]+grain)

    return list_n_grains, list_total_grains
n_days = 108
list_n_grains, list_total_grains = compute_sorori_shinzaemon(n_days=n_days)
print(f'Number of grains on the {n_days} days: {list_n_grains[-1]}')
print(f'The total rice grain after {n_days} days : {list_total_grains[-1]}')

# Plotting the results
plt.style.use("classic")
fig, ax = plt.subplots()
ax.plot(list_n_grains, color='red', lw= 1.5, label='Daily Rice Grains')
ax.plot(list_total_grains, color='green', lw= 1.5, label='Total Daily Rice Grains')
plt.title(f"Daily Rice Grains over {n_days} days")
ax.set_xlabel('Days')
ax.set_ylabel('Number of Rice Grains')
ax.set_xscale('linear')
ax.set_yscale('linear')
ax.grid(True)
ax.legend(loc='upper left')
plt.show()


```
