## Problem 1: Data Acquisition

```python
from sklearn.datasets import load_iris
import pandas as pd
iris = load_iris()
# Create DataFrame x with feature values
X = pd.DataFrame(
    data=iris.data,
    columns=['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
)

# Create DataFrame y with target values(species)
Y = pd.DataFrame(
    data=iris.target,
    columns=['species']
)
```

## Problem 2: Combining Data

```python
df = pd.concat([X, Y], axis=1)
df.head()
```

## Problem 4: Examining the iris dataset itself

El Iris Dataset es uno de los conjuntos de datos más icónicos en el mundo del machine learning y la estadística. Introducido por el biólogo Ronald Fisher en 1936, contiene información sobre tres especies de flores del género Iris (Iris setosa, Iris versicolor e Iris virginica). Cada especie está representada por 50 muestras, con cuatro características medidas por flor:

* Longitud del sépalo (cm)

* Ancho del sépalo (cm)

* Longitud del pétalo (cm)

* Ancho del pétalo (cm)

Características clave:
Tamaño: 150 muestras (50 por especie) y 4 características.

Uso común:

* Clasificación supervisada (aprender a predecir la especie a partir de las medidas).

* Pruebas de algoritmos de clustering (como K-Means).

* Visualización de datos multivariados (gráficos de pares, PCA).

Desafío: Las especies versicolor y virginica son más difíciles de distinguir que la setosa, lo que lo hace ideal para probar modelos.

## Problem 6 Creating a diagram


### Pie chart of species distribution

```python
import matplotlib.pyplot as plt

# Convert numerical species to actual names
species_names = {0: 'setosa', 1: 'versicolor', 2: 'virginica'}
df['species_name'] = df['species'].map(species_names)

# Pie chart of sample counts per label
plt.figure(figsize=(10, 6))
df['species_name'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Distribution of Iris Species')
plt.ylabel('')
plt.show()
```

### Box plot for each feature by species

```python
import seaborn as sns
plt.figure(figsize=(10, 8))
features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']

for i, feature in enumerate(features, 1):
  plt.subplot(2, 2, i)
  sns.boxplot(x='species_name', y=feature, data=df)
  plt.title(f'{feature.title()} Distribution')
plt.tight_layout()
plt.show()
```

### Violin Plots Feature Distribution by Species

```python
plt.figure(figsize=(12, 8))
for i, feature in enumerate(features, 1):
  plt.subplot(2, 2, i)
  sns.violinplot(x='species_name', y=feature, data=df, inner='quartile')
  plt.title(f'{feature.title()} Distribution')
plt.tight_layout()
plt.show()
```

## Problem 7 Confirming the relationship between features

### Scatter plots of all combinations of features

```python
fig, ax = plt.subplots(2, 3, figsize=(15, 10))

fig.suptitle('Scatter plots of all combinations of features')

sns.scatterplot(data=df, x='sepal_length', y='sepal_width', ax=ax[0, 0], hue='species', palette='deep')
sns.scatterplot(data=df, x='sepal_length', y='petal_length', ax=ax[0, 1],hue='species', palette='deep')
sns.scatterplot(data=df, x='sepal_length', y='petal_width', ax=ax[0, 2], hue='species', palette='deep')
sns.scatterplot(data=df, x='sepal_width', y='petal_length', ax=ax[1, 0], hue='species', palette='deep')
sns.scatterplot(data=df, x='sepal_width', y='petal_width', ax=ax[1, 1],  hue='species', palette='deep')
sns.scatterplot(data=df, x='petal_length', y='petal_width', ax=ax[1, 2], hue='species', palette='deep')
sns.set
```

###Create a scatter plot matrix

```python
# Scatter plot matrix
sns.pairplot(df, hue='species', height=2.5)
```

### Correlation coefficient matrix of 4 features

```python
# Correlation matrix
# Correlation matrix
correlation_matrix = df[['sepal_length', 'sepal_width', 'petal_length', 'petal_width','species']].corr()
correlation_matrix
```

### Heat map of the correlation matrix

```python
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', linewidths=0.5, fmt='.2f')
plt.title('Correlation Matrix')
plt.show()
```
