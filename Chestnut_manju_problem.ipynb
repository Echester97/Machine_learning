**PROBLEM 1**

If you sprinkle a single chestnut bun with vine, how long will it take to cover the solar system with chestnut buns?  Create a program to calculate this.

```python
solar_system_volume = 1.77e39 # in cubic meters
chestnut_volume = 1.76e-4 # in cubic meters
doubling_time = 5 # in minutes
increasing_time = 0 # in minutes
increased_chestnut = chestnut_volume
while increased_chestnut <= solar_system_volume:
  increased_chestnut *=2
  increasing_time += doubling_time

print(f'it takes {increasing_time/60} hours to cover the solar system with the chestnut')
```

**PROBLEM 2**

Create a generic program that can be applied to other situations, such as the time it takes for the Tokyo Dome to fill up when you sprinkle a ball with Byvine. Also, please create a notebook that is easy to understand, e.g., display a graph

```python
def time_to_fill_space(initial_volume, space_volume, doubling_time):
   time_minutes = 0
   current_volume = initial_volume
   accumulated_volume = 0
   volume = [0] # List to store volume over time
   times = [0] # List to store time point

   while current_volume < space_volume:
    accumulated_volume += current_volume
    volume.append(accumulated_volume)
    times.append(time_minutes)
    current_volume *= 2
    time_minutes += doubling_time
   return time_minutes - doubling_time, times, volume

# Given data
baseball_ball_volume = 2.13e-4
tokyo_dome_volume = 1.24e6
doubling_time = 5 # in minutes
time_needed, times, volume  = time_to_fill_space(baseball_ball_volume, tokyo_dome_volume, doubling_time)

# Plot the growth curve
plt.plot(times, volume, marker='o', linestyle='-', color='g', label='Chestnut Volume Growth')
plt.scatter(time_needed, tokyo_dome_volume, color='b', label='Tokyo Dome Volume')
plt.axhline(y=tokyo_dome_volume, color='r', linestyle='--')
plt.axvline(x=time_needed, color='r', linestyle='--')
plt.xlabel('Time (minutes)')
plt.ylabel('Volume (cubic meters)')
plt.title('Growth Curve of Volume Over Time')
plt.legend()
plt.grid(True)
plt.show()

print(f'Time to fill up Tokyo Dome: {time_needed} minutes')
print(f'Time to fill up Tokyo Dome: {time_needed/60} hours')
```
